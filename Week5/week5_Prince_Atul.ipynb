# Text Generation using RNN, LSTM, and GRU

## Objective
The objective of this project is to build Deep Learning models using Vanilla RNN, LSTM, and GRU to learn text patterns and generate meaningful text. The performance of all three models is then compared.

## About Text Generation
Text generation is a Natural Language Processing (NLP) task where a model predicts the next word in a sequence. By learning from a text corpus, the model can generate new text that follows similar grammar and context.

- **Vanilla RNN:** Basic recurrent network for sequential data.
- **LSTM:** Handles long-term dependencies using memory cells.
- **GRU:** A simpler and faster alternative to LSTM with comparable performance.

## Import Libraries

Import the required libraries for text preprocessing, building deep learning models, and evaluating performance.

```python
# Numerical operations
import numpy as np

# Measure training time
import time

# Text preprocessing
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Build deep learning models
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, SimpleRNN, LSTM, GRU, Dense

# Split dataset
from sklearn.model_selection import train_test_split

# Display comparison table
import pandas as pd

# Ignore warnings
import warnings
warnings.filterwarnings("ignore")

# Set random seed for reproducibility
np.random.seed(42)
```

## Load Dataset

Load the text corpus and display basic information such as the dataset size and a sample of the text.

```python
# Sample text corpus
text = """
Artificial intelligence is transforming the world.
Deep learning enables computers to learn from data.
Neural networks are inspired by the human brain.
Text generation is an important application of natural language processing.
Recurrent Neural Networks process sequential data.
LSTM models capture long-term dependencies in text.
GRU models are efficient and faster than LSTM in many cases.
Machine learning helps solve real-world problems.
Python is widely used for artificial intelligence and data science.
"""

# Remove extra spaces
text = text.strip()

# Display dataset information
print("Dataset Shape:", (1,))
print("\nSample Text:\n")
print(text[:300])
```

**Explanation:**  
The dataset consists of a small text corpus used to train the text generation models. The models learn word sequences and predict the next word based on the input text.

## Text Preprocessing

The text is converted to lowercase, tokenized, converted into sequences, and padded to prepare it for training the deep learning models.

```python
# Convert text to lowercase
text = text.lower()

# Tokenize the text
tokenizer = Tokenizer()
tokenizer.fit_on_texts([text])

# Vocabulary size
total_words = len(tokenizer.word_index) + 1
print("Vocabulary Size:", total_words)

# Create input sequences
input_sequences = []

for line in text.split("\n"):
    token_list = tokenizer.texts_to_sequences([line])[0]

    for i in range(1, len(token_list)):
        n_gram_sequence = token_list[:i + 1]
        input_sequences.append(n_gram_sequence)

# Pad sequences
max_sequence_len = max(len(seq) for seq in input_sequences)

input_sequences = pad_sequences(
    input_sequences,
    maxlen=max_sequence_len,
    padding="pre"
)

# Split into features and labels
X = input_sequences[:, :-1]
y = input_sequences[:, -1]

print("Input Shape:", X.shape)
print("Output Shape:", y.shape)
print("Maximum Sequence Length:", max_sequence_len)
```

**Explanation:**

- Converted text to lowercase.
- Tokenized the text into words.
- Created the vocabulary.
- Generated input sequences.
- Applied padding so all sequences have the same length.

## Build Vanilla RNN Model

A simple Recurrent Neural Network (RNN) is built using an Embedding layer, a SimpleRNN layer, and a Dense output layer.

```python
# Build Vanilla RNN model
rnn_model = Sequential([
    Embedding(input_dim=total_words, output_dim=64),
    SimpleRNN(64),
    Dense(total_words, activation="softmax")
])

# Compile the model
rnn_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Model summary
rnn_model.summary()

# Train the model
start_time_rnn = time.time()

history_rnn = rnn_model.fit(
    X,
    y,
    epochs=50,
    batch_size=16,
    verbose=1
)

rnn_training_time = time.time() - start_time_rnn

print(f"\nRNN Training Time: {rnn_training_time:.2f} seconds")
```

## Evaluate Vanilla RNN

Evaluate the trained RNN model using loss and accuracy, then generate sample text.

```python
# Evaluate the RNN model
rnn_loss, rnn_accuracy = rnn_model.evaluate(X, y, verbose=0)

print(f"RNN Loss     : {rnn_loss:.4f}")
print(f"RNN Accuracy : {rnn_accuracy:.4f}")
```

```python
# Function to generate text
def generate_text(model, seed_text, next_words):
    output_text = seed_text

    for _ in range(next_words):
        token_list = tokenizer.texts_to_sequences([output_text.lower()])[0]
        token_list = pad_sequences(
            [token_list],
            maxlen=max_sequence_len - 1,
            padding="pre"
        )

        predicted = np.argmax(model.predict(token_list, verbose=0), axis=-1)[0]

        output_word = ""
        for word, index in tokenizer.word_index.items():
            if index == predicted:
                output_word = word
                break

        output_text += " " + output_word

    return output_text
```

```python
# Generate sample text using RNN
print("Generated Text 1:")
print(generate_text(rnn_model, "artificial intelligence", 5))

print("\nGenerated Text 2:")
print(generate_text(rnn_model, "deep learning", 5))

print("\nGenerated Text 3:")
print(generate_text(rnn_model, "machine learning", 5))
```

## Build LSTM Model

Build and train an LSTM model to learn long-term dependencies in the text data.

```python
# Build LSTM model
lstm_model = Sequential([
    Embedding(input_dim=total_words, output_dim=64),
    LSTM(64),
    Dense(total_words, activation="softmax")
])

# Compile the model
lstm_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Display model summary
lstm_model.summary()

# Train the model
start_time_lstm = time.time()

history_lstm = lstm_model.fit(
    X,
    y,
    epochs=50,
    batch_size=16,
    verbose=1
)

lstm_training_time = time.time() - start_time_lstm

print(f"\nLSTM Training Time: {lstm_training_time:.2f} seconds")
```

## Evaluate LSTM

Evaluate the trained LSTM model using loss and accuracy, then generate sample text.

```python
# Evaluate the LSTM model
lstm_loss, lstm_accuracy = lstm_model.evaluate(X, y, verbose=0)

print(f"LSTM Loss     : {lstm_loss:.4f}")
print(f"LSTM Accuracy : {lstm_accuracy:.4f}")
```

```python
# Generate sample text using LSTM
print("Generated Text 1:")
print(generate_text(lstm_model, "artificial intelligence", 5))

print("\nGenerated Text 2:")
print(generate_text(lstm_model, "deep learning", 5))

print("\nGenerated Text 3:")
print(generate_text(lstm_model, "machine learning", 5))
```

## Build GRU Model

Build and train a GRU model for text generation. GRU is simpler than LSTM and often trains faster while maintaining good performance.

```python
# Build GRU model
gru_model = Sequential([
    Embedding(input_dim=total_words, output_dim=64),
    GRU(64),
    Dense(total_words, activation="softmax")
])

# Compile the model
gru_model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Display model summary
gru_model.summary()

# Train the model
start_time_gru = time.time()

history_gru = gru_model.fit(
    X,
    y,
    epochs=50,
    batch_size=16,
    verbose=1
)

gru_training_time = time.time() - start_time_gru

print(f"\nGRU Training Time: {gru_training_time:.2f} seconds")
```

## Evaluate GRU

Evaluate the trained GRU model using loss and accuracy, then generate sample text.

```python
# Evaluate the GRU model
gru_loss, gru_accuracy = gru_model.evaluate(X, y, verbose=0)

print(f"GRU Loss     : {gru_loss:.4f}")
print(f"GRU Accuracy : {gru_accuracy:.4f}")
```

```python
# Generate sample text using GRU
print("Generated Text 1:")
print(generate_text(gru_model, "artificial intelligence", 5))

print("\nGenerated Text 2:")
print(generate_text(gru_model, "deep learning", 5))

print("\nGenerated Text 3:")
print(generate_text(gru_model, "machine learning", 5))
```

## Model Comparison

Compare the performance of the Vanilla RNN, LSTM, and GRU models based on loss, accuracy, training time, and text generation quality.

```python
# Create comparison table
comparison = pd.DataFrame({
    "Model": ["Vanilla RNN", "LSTM", "GRU"],
    "Accuracy": [
        round(rnn_accuracy, 4),
        round(lstm_accuracy, 4),
        round(gru_accuracy, 4)
    ],
    "Loss": [
        round(rnn_loss, 4),
        round(lstm_loss, 4),
        round(gru_loss, 4)
    ],
    "Training Time (s)": [
        round(rnn_training_time, 2),
        round(lstm_training_time, 2),
        round(gru_training_time, 2)
    ]
})

# Add text quality based on accuracy
best_accuracy = max(rnn_accuracy, lstm_accuracy, gru_accuracy)

comparison["Generated Text Quality"] = [
    "Best" if acc == best_accuracy else "Good"
    for acc in comparison["Accuracy"]
]

# Display comparison table
print(comparison)

# Display best model
best_model = comparison.loc[comparison["Accuracy"].idxmax(), "Model"]

print(f"\nBest Performing Model: {best_model}")
```
