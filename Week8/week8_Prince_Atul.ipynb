# Agentic AI Pipeline using a Single Smart Agent

## Objective

The objective of this project is to build a simple Agentic AI pipeline.
The agent understands the user's query, selects the correct tool,
and returns the result in JSON format.

## Required Libraries

This project uses only Python built-in functions.
No external libraries are required.

```python
# No external libraries required
```

## Agentic AI Pipeline

Pipeline Flow:

User Query
      ↓
Single Agent
      ↓
Check Intent
      ↓
Calculator Tool / Keyword Tool / General Response
      ↓
JSON Output

### Step 1

Create a calculator tool that evaluates mathematical expressions.

```python
def calculator(expression):
    """Calculate mathematical expressions."""
    try:
        return str(eval(expression))
    except:
        return "Error in calculation"
```

### Step 2

Create a keyword extraction tool.

```python
def extract_keywords(text):
    """Extract important words."""
    try:
        words = text.split()

        keywords = list(
            set(
                [
                    word.lower().strip(".,!?")
                    for word in words
                    if len(word) > 4
                ]
            )
        )

        return keywords[:5]

    except:
        return []
```

### Step 3

The agent checks the user's query and routes it
to the correct tool.

```python
def agent(query):

    query_lower = query.lower()

    # Calculator Tool
    if "calculate" in query_lower:

        expression = query_lower.replace("calculate", "").strip()

        result = calculator(expression)

        return {
            "type": "calculation",
            "result": result
        }

    # Keyword Tool
    elif "keywords" in query_lower:

        text = query.replace("Extract keywords from", "")
        text = text.replace("extract keywords from", "")

        result = extract_keywords(text)

        return {
            "type": "keywords",
            "result": result
        }

    # General Response
    else:

        return {
            "type": "general",
            "result": "This is a general query. No tool required."
        }
```

### Step 4

Run a few sample queries to verify the pipeline.

```python
queries = [

    "Calculate 20 + 5",

    "Extract keywords from Artificial Intelligence is transforming industries",

    "What is machine learning?"

]

for q in queries:

    print("Query:", q)

    print("Response:", agent(q))

    print("-"*50)
```

## Run the Pipeline

The following cell allows the user to interact with the agent.
Type "exit" to stop.

```python
while True:

    user_input = input("Enter Query : ")

    if user_input.lower() == "exit":
        break

    print(agent(user_input))
```
